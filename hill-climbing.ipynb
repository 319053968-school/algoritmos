```python
class Node:

    def __init__(self, state, parent=None):

        self.state = state
        self.parent = parent

    def path(self):

        camino = []
        nodo = self

        while nodo is not None:

            camino.append(nodo.state)
            nodo = nodo.parent

        camino.reverse()

        return camino

    def __repr__(self):

        return f"Node({self.state})"

    def expand(self, problem):
    
        vecinos = []
    
        for estado in problem.neighbors(self.state):
    
            vecinos.append(
                Node(
                    estado,
                    parent=self
                )
            )
    
        return vecinos
```

```python
class Problem:

    def __init__(self, initial, goal, graph, heuristic):

        self.initial = initial
        self.goal = goal
        self.graph = graph
        self.heuristic = heuristic

    def is_goal(self, node):

        return node.state == self.goal

    def neighbors(self, state):

        return list(self.graph.neighbors(state))

    def h(self, node):

        return self.heuristic[node.state]
```

```python
def hill_climbing(problem):

    current = Node(problem.initial)

    print("INICIO DE HILL CLIMBING")

    while True:

        print(
            f"\nNodo actual: {current.state} "
            f"| h = {problem.h(current)} km"
        )

        if problem.is_goal(current):

            print("\nMeta encontrada:", current.state)

            return current

        neighbors = current.expand(problem)

        if not neighbors:

            print("\nNo hay vecinos disponibles.")

            return current

        print("\nVecinos:")

        for neighbor in neighbors:

            print(
                f"  {neighbor.state} "
                f"| h = {problem.h(neighbor)} km"
            )

        best_neighbor = min(
            neighbors,
            key=problem.h
        )

        print(
            f"\nMejor vecino: {best_neighbor.state} "
            f"| h = {problem.h(best_neighbor)} km"
        )

        if problem.h(best_neighbor) >= problem.h(current):

            print("\nNo existe un vecino que mejore la heurística.")
            print("Hill Climbing se detiene.")

            return current

        print(
            f"Avanza: {current.state} "
            f"-> {best_neighbor.state}"
        )

        current = best_neighbor
```

```python
problema_1 = Problem(
    "Cuatro Caminos",
    "Pantitlán",
    metro_r1,
    heuristica_pantitlan
)

resultado = hill_climbing(problema_1)

print("\n==============================")
print("RESULTADO")
print("==============================")

print(
    "Ruta:",
    " -> ".join(resultado.path())
)

print(
    "Heurística final:",
    problema_1.h(resultado),
    "km"
)
```
